```python
#W4A6
import sys
import math
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, math.isqrt(n) + 1):
        if n % i == 0:
            return False
    return True
for line in sys.stdin:
    parts = line.split()
    if not parts:
        continue
    a, b = map(int, parts)
    tong = 0
    for x in range(a, b + 1):
        if is_prime(x):
            tong += x
    print(tong)
```

```python
#W4A9
import math
n = int(input())
result = []
for i in range(1, math.isqrt(n) + 1):
    sq = i * i
    s = str(sq)
    if len(s) == len(set(s)):
        result.append(s)
if result:
    print(" ".join(result))
else:
    print("(no number)")
```

```python
#W4A10
n = int(input())
memo = {1: 1}
def collatz_length(x):
    if x in memo:
        return memo[x]
    if x % 2 == 0:
        memo[x] = 1 + collatz_length(x // 2)
    else:
        memo[x] = 1 + collatz_length(3 * x + 1)
    return memo[x]
best_x = 1
max_len = 1
for x in range(1, n + 1):
    curr_len = collatz_length(x)
    if curr_len > max_len:
        max_len = curr_len
        best_x = x
print(best_x, max_len)
```

```python
#W4A11
import math
n = int(input())
count = 0
for i in range(1, math.isqrt(n) + 1):
    if n % i == 0:
        if i % 2 == 0:
            count += 1
        other = n // i
        if other != i and other % 2 == 0:
            count += 1
print(count)
```
